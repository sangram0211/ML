```python
import pandas as pd

from mlxtend.preprocessing import TransactionEncoder
from mlxtend.frequent_patterns import apriori, association_rules
```

```python
transactions = [
    ['Milk', 'Bread', 'Butter'],
    ['Bread', 'Butter'],
    ['Milk', 'Bread'],
    ['Milk', 'Bread', 'Butter', 'Eggs'],
    ['Bread', 'Eggs'],
    ['Milk', 'Bread', 'Eggs'],
    ['Milk', 'Butter'],
    ['Bread', 'Butter', 'Eggs'],
    ['Milk', 'Bread', 'Butter'],
    ['Bread', 'Milk']
]

print("Number of Transactions:", len(transactions))
```

```python
te = TransactionEncoder()

te_array = te.fit(transactions).transform(transactions)

df = pd.DataFrame(te_array, columns=te.columns_)

print("--- Transaction Dataset ---")
print(df.astype(int))
```

```python
min_support = 0.30

print("Minimum Support:", min_support)
```

```python
frequent_itemsets = apriori(
    df,
    min_support=min_support,
    use_colnames=True
)

frequent_itemsets['itemset_size'] = frequent_itemsets['itemsets'].apply(len)

print("--- Frequent Itemsets ---")

print(
    frequent_itemsets.sort_values(
        by=['itemset_size', 'support'],
        ascending=[True, False]
    )
)
```

```python
print("Frequent Itemsets:\n")

for index, row in frequent_itemsets.iterrows():
    print(
        "Items:",
        list(row['itemsets']),
        " | Support:",
        round(row['support'], 2)
    )
```

```python
min_confidence = 0.60

print("Minimum Confidence:", min_confidence)
```

```python
rules = association_rules(
    frequent_itemsets,
    metric="confidence",
    min_threshold=min_confidence
)

print("--- Association Rules ---")

print(
    rules[
        ['antecedents',
         'consequents',
         'support',
         'confidence',
         'lift']
    ]
)
```

```python
rules = rules.sort_values(
    by=['lift', 'confidence'],
    ascending=False
)

print("--- Sorted Association Rules ---")

print(
    rules[
        ['antecedents',
         'consequents',
         'support',
         'confidence',
         'lift']
    ]
)
```

```python
strong_rules = rules[
    (rules['confidence'] >= min_confidence) &
    (rules['lift'] > 1)
]

print("--- Strong Positive Association Rules ---")

print(
    strong_rules[
        ['antecedents',
         'consequents',
         'support',
         'confidence',
         'lift']
    ]
)
```

```python
display_rules = strong_rules[
    ['antecedents', 'consequents', 'support', 'confidence', 'lift']
].copy()

display_rules['antecedents'] = display_rules['antecedents'].apply(
    lambda x: ', '.join(x)
)

display_rules['consequents'] = display_rules['consequents'].apply(
    lambda x: ', '.join(x)
)

display_rules['support'] = display_rules['support'].round(3)
display_rules['confidence'] = display_rules['confidence'].round(3)
display_rules['lift'] = display_rules['lift'].round(3)

display(display_rules)
```

```python
best_rule = strong_rules.iloc[0]

print("Best Association Rule")
print("----------------------")
print("If customer buys:", list(best_rule['antecedents']))
print("Then customer may buy:", list(best_rule['consequents']))
print("Support:", round(best_rule['support'], 3))
print("Confidence:", round(best_rule['confidence'], 3))
print("Lift:", round(best_rule['lift'], 3))
```
